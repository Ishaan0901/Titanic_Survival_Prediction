```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
x_train=pd.read_csv('data/x_train.csv',index_col=0)
x_test=pd.read_csv('data/x_test.csv',index_col=0)

print(x_train.shape)
print(x_test.shape)
```

```python
y_train=pd.read_csv('data/y_train.csv',index_col=0).squeeze()
y_test=pd.read_csv('data/y_test.csv',index_col=0).squeeze()

print(len(y_train))
print(len(y_test))
print(type(y_train))
```

```python
model= RandomForestClassifier(random_state=42,max_depth=4)
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
ran_acc=accuracy_score(y_test,y_pred)
ran_acc
```

## Train Accuracy:

```python
train_pred=model.predict(x_train)
```

```python
train_acc=accuracy_score(y_train,train_pred)
train_acc
```

## Feature Importance:

```python
model.feature_importances_
```

```python
df=pd.DataFrame(model.feature_importances_,x_train.columns)
df=df.sort_values(ascending=False,by=0)
df
```

```python
df.plot(kind='barh',legend=False,title='Feature Importance')
```
